## 1. Une première cellule

Trois choses à retenir : le `#` introduit un commentaire, le `=` affecte une valeur, et rien ne s'affiche sans `print()`.

```python
# Le taux de croissance du PIB entre deux annees
pib_2023 = 2822
pib_2024 = 2891

croissance = (pib_2024 - pib_2023) / pib_2023 * 100
print("Le taux de croissance est de", croissance, "%")
```

Le résultat s'affiche avec quinze décimales. C'est normal : Python stocke les nombres à virgule de façon approchée, et il affiche tout ce qu'il a.

Pour un résultat lisible, on arrondit à l'affichage — jamais dans le calcul lui-même.

```python
print("Le taux de croissance est de", round(croissance, 2), "%")
```

## 3. Les listes

Une liste retient un **ordre**. On y accède par la position.

```python
prix = [1.20, 0.95, 2.40, 1.75]

print(prix[0])      # le premier -- on compte a partir de ZERO
print(prix[-1])     # le dernier
print(prix[0:2])    # les deux premiers -- la borne de droite est EXCLUE
```

Deux conventions à intégrer une fois pour toutes :

- l'indexation commence à **zéro** ;
- la borne de droite d'une tranche est **exclue**.

Se tromper sur l'une ou l'autre ne produit pas d'erreur. Cela produit un résultat faux — c'est bien plus ennuyeux.

```python
print(len(prix))    # combien d'elements
print(sum(prix))    # la somme
```

## 4. Les dictionnaires

Là où une liste retient un ordre, un dictionnaire retient une **correspondance** : chaque valeur porte un nom.

C'est la structure d'une ligne de tableau de données — et celle d'un panier de biens.

```python
prix = {"pain": 1.20, "lait": 0.95, "cafe": 2.40}

print(prix["pain"])       # on accede par la cle, pas par la position
print(prix.keys())        # les biens
print(prix.values())      # les prix
```

```python
# Tester la presence d'une cle
print("pain" in prix)
print("beurre" in prix)
```

## 5. Lire un message d'erreur

Que se passe-t-il si l'on demande une clé qui n'existe pas ?

```python
# Cette cellule produit une ERREUR. C'est voulu : lisez-la.
prix["beurre"]
```

Un message d'erreur Python se lit **par la fin** : la dernière ligne dit ce qui ne va pas, celles du dessus disent où.

Ici : `KeyError: 'beurre'` — cette clé n'existe pas.

Les quatre erreurs que vous rencontrerez le plus :

| Erreur | Signification |
|---|---|
| `KeyError` | cette clé n'existe pas |
| `NameError` | cette variable n'a jamais été définie — cellule non exécutée ? |
| `TypeError` | opération impossible entre ces types |
| `FileNotFoundError` | le fichier n'est pas là où vous croyez |

**Une erreur qui s'affiche est une bonne nouvelle.** Ce sont les calculs silencieusement faux qui coûtent cher — comme le prix TTC de la partie 2.

```python
# La methode .get() ne leve PAS d'erreur : elle renvoie None
resultat = prix.get("beurre")
print(resultat)
```

Attention : c'est commode, mais c'est aussi un piège. `.get()` transforme une erreur bruyante en absence silencieuse. À n'utiliser que si l'absence est un cas normal et prévu.

## 6. Application — de combien les prix ont-ils augmenté ?

Deux dictionnaires suffisent à poser le problème économique.

```python
prix_2023 = {"pain": 1.20, "lait": 0.95, "cafe": 2.40}
prix_2024 = {"pain": 1.35, "lait": 1.02, "cafe": 2.28}
```

### 6.1 La variation de chaque prix

```python
var_pain = (prix_2024["pain"] - prix_2023["pain"]) / prix_2023["pain"] * 100
var_lait = (prix_2024["lait"] - prix_2023["lait"]) / prix_2023["lait"] * 100
var_cafe = (prix_2024["cafe"] - prix_2023["cafe"]) / prix_2023["cafe"] * 100

print("pain :", round(var_pain, 2), "%")
print("lait :", round(var_lait, 2), "%")
print("cafe :", round(var_cafe, 2), "%")
```

Notez que le prix du café **baisse**. Un indice de prix n'est donc pas « la hausse des prix » : c'est le résultat d'une agrégation, et certains termes peuvent être négatifs.

### 6.2 Première réponse : la moyenne des variations

```python
moyenne_var = (var_pain + var_lait + var_cafe) / 3
print("Moyenne des variations :", round(moyenne_var, 2), "%")
```

**4,96 %.** Retenez ce chiffre.

Cette moyenne traite le pain, le lait et le café à égalité. Or un ménage n'achète pas les trois dans les mêmes proportions.

### 6.3 Seconde réponse : le coût du panier

On introduit les quantités consommées sur l'année.

```python
quantites = {"pain": 120, "lait": 90, "cafe": 24}

facture_2023 = (prix_2023["pain"] * quantites["pain"]
              + prix_2023["lait"] * quantites["lait"]
              + prix_2023["cafe"] * quantites["cafe"])

facture_2024 = (prix_2024["pain"] * quantites["pain"]
              + prix_2024["lait"] * quantites["lait"]
              + prix_2024["cafe"] * quantites["cafe"])

print("Facture 2023 :", round(facture_2023, 2), "euros")
print("Facture 2024 :", round(facture_2024, 2), "euros")
```

```python
var_facture = (facture_2024 - facture_2023) / facture_2023 * 100
print("Variation de la facture :", round(var_facture, 2), "%")
```

**7,46 %**, contre 4,96 % pour la moyenne simple. Deux points et demi d'écart, sur les mêmes prix.

D'où vient la différence ?

### 6.4 Pourquoi cet écart : les parts budgétaires

```python
for bien in prix_2023:
    part = prix_2023[bien] * quantites[bien] / facture_2023 * 100
    print(bien, ":", round(part, 1), "% du budget")
```

Le pain pèse la moitié du budget et son prix augmente de 12,5 %. Le café n'en pèse que 20 % et son prix baisse.

La moyenne simple donnait le même poids aux trois. Le panier donne à chacun son poids réel.

```python
# Verification : le panier EST une moyenne ponderee des variations
moyenne_ponderee = (
      var_pain * prix_2023["pain"] * quantites["pain"] / facture_2023
    + var_lait * prix_2023["lait"] * quantites["lait"] / facture_2023
    + var_cafe * prix_2023["cafe"] * quantites["cafe"] / facture_2023
)

print("Moyenne ponderee des variations :", round(moyenne_ponderee, 4), "%")
print("Variation de la facture         :", round(var_facture, 4), "%")
```

Les deux nombres coïncident. Ce n'est pas un hasard : **un indice de prix est une moyenne des variations, pondérée par les parts budgétaires**.

Vous venez de calculer un indice de Laspeyres, sans le savoir : les quantités retenues sont celles de l'année de base.

## 7. La question ouverte

Rien ne nous obligeait à retenir les quantités de 2023. Nous aurions pu prendre celles de 2024.

| | Quantités utilisées | La question posée |
|---|---|---|
| **Laspeyres** | année de base | combien coûterait aujourd'hui le panier d'hier ? |
| **Paasche** | année courante | combien coûtait hier le panier d'aujourd'hui ? |

Les deux mesurent « l'inflation » et ne donnent pas le même chiffre. L'écart n'est pas une erreur : les ménages **substituent** quand les prix relatifs changent.

C'est l'objet du TD 1, sur les données réelles d'Eurostat.
