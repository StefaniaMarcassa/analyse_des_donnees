### Fusionner deux tableaux

`pd.merge` associe les lignes de deux tableaux qui partagent une **clé**.

```python
f = gauche.merge(droite, on="cle", how="left", indicator=True)
```

- `how="left"` garde toutes les lignes de gauche, qu'elles trouvent un partenaire ou non ;
- `indicator=True` ajoute une colonne `_merge` : `both` si la ligne s'est appariée, `left_only` sinon.

**Comptez toujours les lignes avant et après.**
